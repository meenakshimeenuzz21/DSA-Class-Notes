```python
def run_etl_pipeline():

    df = extract()

    df = df.drop_duplicates()

    df = df.fillna("Unknown")

    df["City"] = df["City"].str.title()

    df.to_csv("reports/cleaned_dataset.csv",index=False)

    df.head(10).to_csv("reports/top10_records.csv",index=False)

    df.describe(include="all").to_csv("reports/summary_statistics.csv")

    df.to_csv("output/final_dataset.csv",index=False)

    print("ETL Pipeline Completed")
```

```python
import os
os.makedirs("output", exist_ok=True)
run_etl_pipeline()
```
