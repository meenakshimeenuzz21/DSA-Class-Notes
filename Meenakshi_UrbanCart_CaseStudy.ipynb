**Module 5**

**Business Functions & Reporting**

Function 1: Get Product Price

```python
def get_product_price(product_id, products_list):

    for product in products_list:

        if product["Product_ID"] == product_id:
            return product["Price"]

    return None


print("Price of P08:", get_product_price("P08", products))
```

Function 2: Get Customer Name

```python
def get_customer_name(customer_id, customers_list):

    for customer in customers_list:

        if customer["Customer_ID"] == customer_id:
            return customer["Name"]

    return None


print("Customer C01:", get_customer_name("C01", customers))
```

Function 3: Calculate Order Value

```python
def calculate_order_value(price, quantity):

    return price * quantity


print("Order Value:", calculate_order_value(1999, 2))
```

Function 4: Classify Order

```python
def classify_order(value, high_limit=3000, low_limit=1000):

    if value > high_limit:
        return "High"

    elif value >= low_limit:
        return "Medium"

    else:
        return "Low"


print("3500:", classify_order(3500))
print("2000:", classify_order(2000))
print("500:", classify_order(500))
```

Function 5: Generate Invoice

```python
def generate_invoice(order, products_list, customers_list):

    price = get_product_price(
        order["Product_ID"],
        products_list
    )

    customer_name = get_customer_name(
        order["Customer_ID"],
        customers_list
    )

    order_value = calculate_order_value(
        price,
        order["Quantity"]
    )

    product_name = ""

    for product in products_list:

        if product["Product_ID"] == order["Product_ID"]:

            product_name = product["Name"]
            break

    invoice = f"""
========================================
           URBANCART INVOICE
========================================
Order ID      : {order["Order_ID"]}
Customer      : {customer_name}
Product       : {product_name}
Quantity      : {order["Quantity"]}
Price         : ₹{price}
Order Value   : ₹{order_value}
Payment Mode  : {order["Payment_Mode"]}
Status        : {order["Status"]}
========================================
"""

    return invoice


print(generate_invoice(orders[0], products, customers))
```

Function 6: Calculate Total Revenue

```python
def calculate_total_revenue(orders_list, products_list):

    total_revenue = 0

    for order in orders_list:

        # Skip cancelled orders
        if order["Status"] == "Cancelled":
            continue

        # Find the product price
        for product in products_list:

            if product["Product_ID"] == order["Product_ID"]:

                order_value = product["Price"] * order["Quantity"]

                total_revenue += order_value

                break

    return total_revenue


# Call the function
function_revenue = calculate_total_revenue(orders, products)

print("Total Revenue using function:", function_revenue)

# Compare with Module 4 answer
print("Module 4 Total Revenue:", total_revenue)

if function_revenue == total_revenue:
    print("Confirmed: Both revenue values match.")
else:
    print("Revenue values do not match.")
```

Function 7: Top Customer

```python
def top_customer_by_spend(
    orders_list,
    products_list,
    customers_list
):

    customer_spending = {}

    for order in orders_list:

        if order["Status"] != "Delivered":
            continue

        price = get_product_price(
            order["Product_ID"],
            products_list
        )

        order_value = calculate_order_value(
            price,
            order["Quantity"]
        )

        customer_id = order["Customer_ID"]

        if customer_id not in customer_spending:
            customer_spending[customer_id] = 0

        customer_spending[customer_id] += order_value

    top_customer_id = None
    highest_spend = 0

    for customer_id in customer_spending:

        if customer_spending[customer_id] > highest_spend:

            highest_spend = customer_spending[customer_id]
            top_customer_id = customer_id

    return get_customer_name(
        top_customer_id,
        customers_list
    )


print(
    "Top Customer:",
    top_customer_by_spend(
        orders,
        products,
        customers
    )
)
```

Bonus *args

```python
def apply_offer(price, *discounts):

    final_price = price

    for discount in discounts:

        final_price = final_price - (
            final_price * discount / 100
        )

    return final_price


print("Original Price: ₹1000")

print(
    "After 10% discount:",
    apply_offer(1000, 10)
)

print(
    "After 10% and 5% discounts:",
    apply_offer(1000, 10, 5)
)
```

```python
delivered_total = 0
delivered_orders = 0

for order in orders:

    if order["Status"] != "Delivered":
        continue

    price = get_product_price(
        order["Product_ID"],
        products
    )

    order_value = calculate_order_value(
        price,
        order["Quantity"]
    )

    delivered_total += order_value
    delivered_orders += 1


if delivered_orders > 0:

    average_order_value = delivered_total / delivered_orders

else:

    average_order_value = 0


print("Average Order Value: ₹", average_order_value)
```

Most Popular Category

```python
category_quantity = {}

for order in orders:

    # Consider only delivered orders
    if order["Status"] != "Delivered":
        continue

    # Find the product for the order
    for product in products:

        if product["Product_ID"] == order["Product_ID"]:

            category = product["Category"]

            # Create category if it does not exist
            if category not in category_quantity:
                category_quantity[category] = 0

            # Add the quantity ordered
            category_quantity[category] += order["Quantity"]

            break


# Find the category with the highest quantity
most_popular_category = None
highest_quantity = 0

for category in category_quantity:

    if category_quantity[category] > highest_quantity:
        highest_quantity = category_quantity[category]
        most_popular_category = category


print("Most popular product category:", most_popular_category)
print("Quantity sold:", highest_quantity)
```
