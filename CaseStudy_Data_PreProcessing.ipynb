**Duplicate Removal**

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
duplicate_columns = []

for i in range(len(df.columns)):
    for j in range(i+1, len(df.columns)):
        if df.iloc[:, i].equals(df.iloc[:, j]):
            duplicate_columns.append(df.columns[j])

duplicate_columns

df = df.drop(columns=duplicate_columns)
```

**Handling Missing Values**

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
df
```

```python
cat_cols = df.select_dtypes(include='object').columns
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
num_imputer = SimpleImputer(strategy='mean')

df[num_cols] = num_imputer.fit_transform(df[num_cols])
df
```

```python
cat_imputer = SimpleImputer(strategy='most_frequent')
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
df


```

```python
df.isnull().sum()
```

Missing values were identified using df.isnull().sum(). All numerical features with missing values were handled using mean imputation by replacing missing values with the mean of their respective columns. The categorical feature contained no missing values; therefore, no imputation was required.

**Scaling Numerical Variables**

```python
df = pd.read_csv("/content/House_Pricing.csv")
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
target = "Sale Price"

scale_cols = [col for col in num_cols if col != target]

scaler = StandardScaler()

df[scale_cols] = scaler.fit_transform(df[scale_cols])

df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[num_cols] = scaler.fit_transform(df[num_cols])

df.head()
```

**Encoding Categorical Variables**

```python
cat_cols = df.select_dtypes(include='object').columns

cat_cols
```

```python
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)

df.head()
```

**Outlier Removal**

```python
for col in scale_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

df.shape
```

```python
num_cols = df.select_dtypes(include=np.number).columns

num_cols = num_cols.drop('Sale Price')

for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

df.shape
```

**Train-Test Split**

```python
X = df.drop('Sale Price', axis=1)

y = df['Sale Price']
```
