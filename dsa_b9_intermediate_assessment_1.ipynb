## Challenge 1: Word Frequency Analyzer
Write a function `top_k_words(filename: str, k: int) -> list` that returns the top k frequent words.
- Ignore case and punctuation.
- Sort by frequency, then alphabetically.


```python
import re
from collections import Counter

def top_k_words(filename: str, k: int) -> list:
    with open(filename, "r", encoding="utf-8") as f:
        text = f.read()

    # Ignore case and punctuation, extract words
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    # Count word frequencies
    word_counts = Counter(words)

    # Sort by frequency descending, then alphabetically ascending
    sorted_words = sorted(word_counts.items(), key=lambda x: (-x[1], x[0]))

    return [word for word, count in sorted_words[:k]]
    pass

# Sample Test (create a sample file)
with open("sample.txt", "w") as f:
    f.write("Data science is evolving. Python is versatile. Learning never stops.")

print(top_k_words("sample.txt", 3))  #expected ["is", "data", "science"]
```

## Challenge 2 : Analyze daily expenses and extract insights.
User enters:

Number of transactions
Category and amount

Food 200
Travel 500
Food 300
Shopping 700

### Output
* Total expense
* Highest expense category
* Category-wise total

```python
def analyze_expenses():
    n = int(input("Enter number of transactions: "))

    expenses = {}

    total_food = 0
    total_travel = 0
    total_shopping = 0

    for _ in range(n):
        category, amount = input("Enter the category and value: ").split()
        amount = float(amount)

        if category == "food":
            total_food += amount

        elif category == "travel":
            total_travel += amount

        elif category == "shopping":
            total_shopping += amount

        else:
            print("Invalid Category")

    # Calculate total expense
    total = total_food + total_travel + total_shopping

    # Find highest expense category
    total1 = [total_food, total_travel, total_shopping]

    if max(total1) == total_food:
        highest_category = "food"
    elif max(total1) == total_travel:
        highest_category = "travel"
    else:
        highest_category = "shopping"

    # Store category-wise total
    expenses = {
        "Food": total_food,
        "Travel": total_travel,
        "Shopping": total_shopping
    }

    print("Total Expense:", total)
    print("Highest Category:", highest_category)
    print("Category-wise:", expenses)


analyze_expenses()
```
