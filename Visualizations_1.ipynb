```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="species", data=penguins)
plt.title("Distribution of Penguin Species")
plt.show()
```

INSIGHTS

.Adelie penguins are the most common species.

.Chinstrap penguins are the least common.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="species", y="flipper_length_mm", data=penguins)
plt.title("Flipper Length by Species")
plt.show()
```

INSIGHTS

.Gentoo penguins have the longest flippers.

.Adelie penguins have the shortest flippers.

```python
plt.figure(figsize=(6,4))
sns.histplot(penguins["body_mass_g"], bins=20, kde=True)
plt.title("Body Mass Distribution")
plt.xlabel("Body Mass (g)")
plt.show()
```

INSIGHTS

.Most penguins weigh between 3500 g and 5000 g.

.Very heavy penguins are less common.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="bill_length_mm", y="bill_depth_mm",hue="species", data=penguins)
plt.title("Bill Length vs Bill Depth")
plt.show()
```

INSIGHTS

.The three species form distinct clusters.

.Bill measurements help differentiate species.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="sex", y="body_mass_g", data=penguins)
plt.title("Body Mass by Sex")
plt.show()
```

INSIGHTS

.Male penguins generally weigh more than female penguins.

.There is some overlap between the groups.

```python
plt.figure(figsize=(7,5))

numeric_df = penguins.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

INSIGHTS

.Body mass is positively correlated with flipper length.

.Bill length also shows a positive relationship with body mass.

.Bill depth has a weaker correlation with the other measurements.

```python
sns.pairplot(penguins, hue="species")
plt.show()
```

INSIGHTS

.Gentoo penguins are clearly separated from the other species.

.Bill length, bill depth, flipper length, and body mass help distinguish species.

```python
plt.figure(figsize=(6,4))
sns.barplot(x="species", y="flipper_length_mm", data=penguins)
plt.title("Average Flipper Length by Species")
plt.show()
```

INSIGHTS

.Gentoo penguins have the longest average flippers.

.Adelie penguins have the shortest average flippers.

```python
plt.figure(figsize=(6,4))
sns.kdeplot(data=penguins,x="body_mass_g",hue="species",fill=True)
plt.title("Density of Body Mass by Species")
plt.show()
```

INSIGHTS

.Gentoo penguins have a body mass distribution shifted toward higher values.

.Adelie and Chinstrap distributions overlap more, indicating similar body masses.

```python
plt.figure(figsize=(6,4))
sns.stripplot(x="species", y="body_mass_g", data=penguins)
plt.title("Body Mass by Species (Strip Plot)")
plt.show()
```

INSIGHTS

.Displays every penguin as an individual point.

.Gentoo penguins generally have higher body mass.

```python
sns.jointplot(data=penguins,x="bill_length_mm",y="bill_depth_mm",kind="scatter")
plt.show()
```

INSIGHTS

.Shows the relationship between bill length and bill depth.

.Histograms on the sides display the distribution of each variable.

```python
plt.figure(figsize=(6,4))
sns.pointplot(x="species", y="body_mass_g", data=penguins)
plt.title("Average Body Mass by Species")
plt.show()
```

INSIGHTS

.Gentoo penguins have the highest average body mass.

.Adelie penguins have the lowest average body mass.

```python
plt.figure(figsize=(6,4))
sns.regplot(x="flipper_length_mm", y="body_mass_g", data=penguins)
plt.title("Flipper Length vs Body Mass")
plt.show()
```

INSIGHTS

.Shows a positive relationship.

.Penguins with longer flippers tend to have higher body mass.
